```python
import matplotlib.pyplot as plt

# Generamos un gráfico de dispersión simple
x = [1, 2, 3, 4, 5]
y = [10, 24, 36, 40, 52]

plt.scatter(x, y, color='cyan', edgecolors='black')
plt.title('Prueba de Telemetría')
plt.xlabel('Eje X')
plt.ylabel('Eje Y')
plt.grid(True, linestyle='--', alpha=0.6)
plt.show()
```
